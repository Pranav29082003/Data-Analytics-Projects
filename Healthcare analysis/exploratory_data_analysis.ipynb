```python
import pandas as pd
import numpy as np 

df=pd.read_csv('healthcare_dataset.csv')
df.head(10)
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
##Create a new column Age group

labels=['Young Adult','Adult','Middle-Aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)

df[['age','age_group']].head(10)
```

```python
##Convert the name Column to Title case and remove the leading/trailing whitespace

df['name']=df['name'].str.title()
df['name']=df['name'].str.strip()

df['name'].head(10)
```

```python
df.columns
```

```python
##Check for 5 most frequent patients in a row
df['name'].value_counts().head(5)
```

```python
##Create unique ID for each patient 
df['patient_id']=df['name'].astype('category').cat.codes
```

```python
df.head(5)
```

```python
##Drop column name for privacy of the patients
df=df.drop(columns=['name'])
```

```python
df.head(5)
```

```python
##Check the data type for date columns

df[['date_of_admission','discharge_date']].dtypes
```

```python
##Convert to Datetime Format

df['date_of_admission']=pd.to_datetime(df['date_of_admission'])
df['discharge_date']=pd.to_datetime(df['discharge_date'])

df[['date_of_admission','discharge_date']].dtypes
```

```python
##Calculate the differnce in days of admission and discharge for that created a new column 'length_of_stays'

df['length_of_stays']=(df['discharge_date']-df['date_of_admission']).dt.days

df['length_of_stays']<0
```

```python
df.columns
```

```python
df[['date_of_admission','discharge_date','length_of_stays']].head(5)
```

```python
##Create a Age distribution Histogram

import matplotlib.pyplot as plt 
import seaborn as sns

plt.figure(figsize=(10,6))
sns.histplot(df['age'],bins=20,kde=True,color='skyblue')
plt.title('Patient Age Distribution',fontsize=15)
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.show()
```

```python
##Top 10 Most common Diseases using a Horizontal Bar Plot

plt.figure(figsize=(10,8))
top_10_disease=df['medical_condition'].value_counts().head(10)

sns.barplot(x=top_10_disease.values,y=top_10_disease.index,palette='magma')
plt.title('Top 10 Common Diseases',fontsize=15)
plt.xlabel('Number of Cases')
plt.ylabel('Disease Name')
plt.show()
```

```python
## Create Pie Chart Showing Doctors Workload

top_doctors=df['doctor'].value_counts().head(5)
plt.figure(figsize=(8,8))
plt.pie(x=top_doctors,labels=top_doctors.index,autopct='%1.1f%%',startangle=140,colors=sns.color_palette('pastel'))
plt.title('Distribution of Patients Among Top Doctors',fontsize=15)
plt.show()
```

```python
##drop columns which are unecessary

df=df.drop(columns=['blood_type','test_results','room_number'])
```
